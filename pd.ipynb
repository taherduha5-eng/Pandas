```python
# Handling Missing Values The dataset contains missing values in the Salary and Bonus columns.
# • Fill the missing Salary values with the median salary of the entire company.
# • Fill the missing bouns values with 0
```

```python
median_salary = df["Salary"].median()
```

```python
df["Salary"].fillna(median_salary,  inplace=True)
```

```python
df["Bonus"].fillna(0, inplace=True)
```

```python
df.isnull().sum()
```

```python
# Data Cleaning: The Employee_Name column is in the format "FirstName LastName",
# ". Split this column into two new columns: First_Name and Last_Name,
# then delete the original Employee_Name column.
```

```python
import pandas as pd

df[['First_Name', 'Last_Name']] = df['Employee_Name'].str.split(' ', expand=True)


df.drop('Employee_Name', axis=1, inplace=True)
df.head()
```

```python
# Filtering :Create a new DataFrame called Senior_IT that contains only employees from the 'IT' department who have more than 8 years of experience.
```

```python
Senior_IT = df[(df["Department"] == "IT") & (df["Years_Experience"] > 8)]

Senior_IT
```

```python
# 5. Performance Mapping Create a new column called Performance_Level based on the Evaluation_Score (1-100):
# • Score > 80: "High"
# • Score 50-80: "Medium"
# • Score < 50: "Low"
```

```python
def performance_level(score):
    if score>80:
        return"high"
    elif score<80:
        return"medium"
    else:
        return"low"
df["Performance_Level"] = df["Evaluation_Score"].apply(performance_level)
    
df.head()
    
```

```python
# Calculate the total salary expense and the average evaluation score for each Department
```

```python
depart_summary = df.groupby("Department").agg({
    "Salary":"sum",
    "Evaluation_Score": "mean"})
depart_summary
```

```python
# 7. Find the top 5 highest-paid employees in the company. Print only their names and salaries.
```

```python
top_5 = df.sort_values(by="Salary", ascending=False)

top_5[["First_Name", "Last_Name", "Salary"]].head(5)
```

```python

#  The company is giving a 10% raise to everyone in the 'Sales' department. Update the Salary column for those employees only.
```

```python
df.loc[df["Department"] == "Sales", "Salary"] *= 1.10
```
